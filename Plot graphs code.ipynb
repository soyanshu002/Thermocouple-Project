```python
import pandas as pd

# Load data
df = pd.read_csv("Book2.csv")

# Convert date column
df["MEASUREDATE"] = pd.to_datetime(df["MEASUREDATE"], dayfirst=True)

# Separate thermocouples
tc8 = df[df["Sl No."] == 8][["MEASUREDATE", "VALUE"]].rename(columns={"VALUE": "TC8"})
tc57 = df[df["Sl No."] == 57][["MEASUREDATE", "VALUE"]].rename(columns={"VALUE": "TC57"})
tc154 = df[df["Sl No."] == 154][["MEASUREDATE", "VALUE"]].rename(columns={"VALUE": "TC154"})

# Merge by date
merged = tc8.merge(tc57, on="MEASUREDATE").merge(tc154, on="MEASUREDATE")

# Distances (mm)
d_8_57 = 450
d_57_154 = 1000

# Calculate gradients
merged["Gradient_8_57 (C/mm)"] = (merged["TC57"] - merged["TC8"]) / d_8_57
merged["Gradient_57_154 (C/mm)"] = (merged["TC154"] - merged["TC57"]) / d_57_154

# Show result
print(merged)

# Save output
merged.to_csv("temperature_gradient_output.csv", index=False)
```

```python
import pandas as pd

# Read the CSV file
df = pd.read_csv("Book2.csv")

# Convert date column to proper datetime format
df["MEASUREDATE"] = pd.to_datetime(df["MEASUREDATE"], dayfirst=True)

# Pivot table to align thermocouples by date
pivot = df.pivot(index="MEASUREDATE", columns="Sl No.", values="VALUE")

# Distances (mm)
d_8_57 = 450
d_57_154 = 1000

# Calculate gradients
pivot["Gradient_8_57_C_per_mm"] = (pivot[57] - pivot[8]) / d_8_57
pivot["Gradient_57_154_C_per_mm"] = (pivot[154] - pivot[57]) / d_57_154

# Keep only required columns
result = pivot[[8, 57, 154, "Gradient_8_57_C_per_mm", "Gradient_57_154_C_per_mm"]]

# Reset index so date becomes a column again
result = result.reset_index()

# Save output
result.to_csv("Thermocouple_Temperature_Gradient.csv", index=False)

print("File saved as Thermocouple_Temperature_Gradient.csv")
```
